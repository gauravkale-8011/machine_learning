```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df =pd.read_csv("placement_.csv")
df.head()
```

```python
x=df.iloc[:,0].values.reshape(-1,1)
y=df.iloc[:,1].values.reshape(-1,1)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
x_train.shape[0]
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test.reshape(-1,1))
```

```python
plt.scatter(df["cgpa"],df["package"])
plt.plot(x_train, (lr.predict(x_train).reshape(-1,1)),color="red")
plt.xlabel("cgpa")
plt.ylabel("package")
plt.show()
```

```python
# now we will calculate the scores through all metrics
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

- 1)   mean absolute error

```python
print("error",mean_absolute_error(y_pred,y_test))
```

- 2)   mean squared error

```python
print("error",mean_squared_error(y_pred,y_test))
```

- 3)   root mean squared error

```python
print("error",np.sqrt(mean_squared_error(y_pred,y_test)))
```

- 4)   r2_score

```python
print("var",r2_score(y_test,y_pred))
```

- 5)   adjusted r2 score

```python
r2=r2_score(y_test,y_pred)
n=x_test.shape[0]
1-(((1-r2)*(n-1))/((n-1-1)))
```
