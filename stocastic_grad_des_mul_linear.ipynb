### Now we will make our class for stocastic gradient descent

```python
class SGDregressor:
    def __init__(self,epochs=100,lr=0.01):
        self.lr=lr
        self.epochs=100
        self.coef=np.ones(x_train.shape[1])
        self.intercept=0

    def fit(self,x_train,y_train):
        for i in range(self.epochs):
            for j in range(x_train.shape[0]):
                
                idx=np.random.randint(0,x_train.shape[0])
                y_hat=np.dot(self.coef,x_train[idx]) + self.intercept
                intercept_der=-2 * (y_train[idx] - y_hat)
                self.intercept=self.intercept - (self.lr * intercept_der )
                coef_der = -2 * np.dot((y_train[idx] - y_hat),x_train[idx])
                self.coef=self.coef- (self.lr * coef_der)
        print(self.coef,self.intercept)

    def predict(self,x_test):
        y_pred=self.intercept + np.dot(x_test,self.coef) 
        return y_pred
```

```python
sgd=SGDregressor()
sgd.fit(x_train,y_train)
```

```python
y_pred=sgd.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```
